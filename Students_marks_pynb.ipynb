```python
import numpy as np

# 10 students marks
marks = np.array([85, 92, 78, 95, 60, 88, 73, 91, 55, 82])
names = ['Raj','Priya','Kumar','Anitha','Bala',
         'Mani','Selvi','Karthik','Ravi','Deepa']

print("=== Student Marks Analysis ===")
print("Average:", marks.mean())
print("Highest:", marks.max())
print("Lowest:", marks.min())
print("Std Dev:", marks.std())

# Pass / Fail
print("\nPass:", marks[marks >= 50].size, "students")
print("Fail:", marks[marks < 50].size, "students")

# Topper
topper_index = marks.argmax()
print("\nTopper:", names[topper_index], "—", marks.max())
# Grade assign
grades = []
for mark in marks:
    if mark >= 90:
        grades.append('A+')
    elif mark >= 80:
        grades.append('A')
    elif mark >= 70:
        grades.append('B')
    else:
        grades.append('C')

for i in range(len(names)):
    print(f"{names[i]}: {marks[i]} — {grades[i]}")
```
